```python
# 6.5 Tugas Mandiri 


#   Bagian A dan B 

skor_nobita = 0
def update_skor(nilai):
    global skor_nobita

    skor_nobita += nilai

#   Bagian C

def belajar(jam_belajar):
    if jam_belajar < 1:
        print('Nobita hanya membuka buku sebentar')
        return 50

    elif 1 <= jam_belajar < 3:
        print('Nobita belajar secukupnya')
        return 25

    else:
        print('Nobita belajar dengan rajin hari ini')
        return 10

def tidur(jam_tidur):
    if jam_tidur < 1:
        print('Nobita kurang tidur')
        return -10

    elif 1 <= jam_tidur <= 2:
        print('Waktu tidur Nobita seimbang')
        return 20

    else:
        print('Nobita tidur terlalu lama')
        return -20

def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print('Nobita bermain secukupnya')
        return 15

    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print('Nobita masih bisa mengatur waktu dengan baik')
        return 5

    else:
        print('Nobita kebanyakan bermain')
        return -30

def menyontek():
    print('Nobita ketahuan menyontek')
    return -50



#   Bagian D

def jalankan_aktivitas(pilihan):
    
    if pilihan == 1:
        jam_belajar = float(input('Berapa jam Nobita belajar hari ini?'))

        skor = belajar(jam_belajar)

        update_skor(skor)

    elif pilihan == 2:
        jam_tidur = float(input('Berapa jam Nobita tidur siang hari ini?'))

        skor = tidur(jam_tidur)

        update_skor(skor)

    elif pilihan == 3:
        jumlah_gadget = float(input('Berapa gadget yang digunakan Nobita?'))

        waktu_main = float(input('Berapa jam Nobita bermain gadget?'))

        skor = main_bersama_doraemon(jumlah_gadget, waktu_main)

        update_skor(skor)

    elif pilihan == 4:
        skor = menyontek()

        update_skor(skor)

    else:
        print('Pilihan tidak valid, aktivitas hanya ada 4 pilihan')

        
#   Bagian E

def evaluasi_akhir(skor):

    if skor >= 80:
        return 'Nobita termasuk Siswa Teladan'

    elif 50 <= skor < 80:
        return 'Nobita termasuk Siswa Cukup Rajin'

    else:
        return 'Nobita termasuk Siswa Pemalas'



def menu_utama():
    print('-'*70)
    print(f'{'=== PROGRAM PENINGKAT DISIPLIN NOBITA ===':^70}')
    print('-'*70)
    
    print(f'{'1. Belajar'}')
    print(f'{'2. Tidur'}')
    print(f'{'3. Main bersama Doraemon'}')
    print(f'{'4. Menyontek saat ujian'}')

    
    print('-'*70)

    pilihan = int(input('Pilih aktivitas (1 - 4):'))
    return pilihan

if __name__ == '__main__':
    skor_nobita = 0
    while True:
        pilihan_user = menu_utama()
        jalankan_aktivitas(pilihan_user)

        lanjut = input('\n Tambah aktivitas lain (y/n):')
        if lanjut != 'y':
            break


print('-'*70)

print(f'{f'Total Skor Nobita : {skor_nobita}':^70}')
print(f'{f'Hasil Evaluasi Akhir Nobita : {skor_nobita}':^70}')

print('-'*70)





    
```
