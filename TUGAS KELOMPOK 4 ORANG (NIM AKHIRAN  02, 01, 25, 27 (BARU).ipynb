```python
#   Soal Nomor 1
print('Dikerjakan oleh Ayushita Hanny Prameswari')
print('-'*50)

#   Data Celcius
celcius = 60

#   Perhitungan konversi suhu celcius ke fahrenheit
fahrenheit = (celcius * 9 / 5) + 32

#   Hasil
print(f'Hasil Konversi suhu celcius ke fahrenheit =', fahrenheit)
print('-'*50)
```
