```python
#   Tugas Mandiri 5

judul = 'Data Mahasiswa'
print('-'*40)
print(f'{judul:^40}')

nama = input('Masukkan nama')
nim = input('Masukkan nim')
prodi = input('Masukkan prodi')

print('-'*40)
print('Nama  :', nama)
print('NIM   :', nim)
print('Prodi :', prodi)
print('-'*40)
```

```python
#   Mencatat Pengeluaran Harian Nobita 

hari = 1
saldo = 50000
no = 1
# Perulangan berjalan selama hari <= 30
while hari <= 30:
    pengeluaran = int(input(f'{no}. Pengeluaran hari ke-{hari}:'))
    print(f"{no}. Pengeluaran hari ke-{hari} : Rp.{pengeluaran:,.0f}")
    no += 1
    saldo -= pengeluaran


    
    # 3. Cek apakah saldo menjadi minus (< 0)
    if saldo < 0:
        print(f"{no}. Saldo habis di hari ke-{hari}. Doraemon menghentikan pencatatan!")
        break

    # 4. Jika saldo masih mencukupi (>= 0), tampilkan sisa saldo lalu tambah nilai hari
    else:
        print(f"{no}. Sisa saldo hari ke-{hari}: Rp.{saldo:,.0f}")
        no += 1
        hari += 1
```

```python
#   Menganalisis Pengeluaran Mingguan Nobita

pengeluaran_mingguan = [20000, 0, 15000, 0]
no = 1
for i in range (4):
    if pengeluaran_mingguan [i] == 0:
        no += 1
        print(f'{no}. Minggu ke-{i+1}: Nobita tidak jajan, analisis dilewatkan')
        no += 1
        continue

    print(f'{no}. Minggu ke-{i+1}: Total Pengeluaran : Rp{pengeluaran_mingguan[i]:,.0f}')

else:
    print(f'{no}. Analisis mingguan selesai.')
    no += 1
```

```python
#   Menyusun laporan bulanan Nobita

no = 0
for bulan in range (1, 4):
    no += 1
    print(f'\n{no}. Bulan ke-{bulan}')
    total_bulan = 0

    
    for minggu in range (1, 5):
        
        pengeluaran = int(input(f' Minggu ke-{minggu}, pengeluaran : Rp. '))
        print(f'Pengeluaran minggu ke-{minggu}: Rp. {pengeluaran}')
        
        if pengeluaran > 20000:
            print('Pengeluaran terlalu besar! Pencatatan bulan ini dihentikan')
            break
        total_bulan += pengeluaran
    
    else:
        print(f'Total pengeluaran bulan ke-{bulan}: Rp. {total_bulan}')

print(sep='.....')
print('-'*60)
ending = 'Penyusun Laporan Pencatatan Bulan Nobita selesai.'
print(f'{ending:^60}')
print('-'*60)
```

```python
#   Evaluasi Akhir Tahun

saldo_awal = 300000
saldo = saldo_awal

no = 0
for bulan in range (11, 13):
    no += 1
    print(f'\n{no}. Bulan ke-{bulan}')
    
    minggu = 1
    while minggu <= 4:
        libur = input(f'Masukkan apakah minggu tersebut libur (y/n)?')
        if libur == 'y':
            print('Minggu ini dilewati (continue).')
            minggu += 1
            continue

        pengeluaran = int(input('Masukkan pengeluaran minggu ini : Rp'))
        saldo -= pengeluaran

        if saldo < 0:
            print('Saldo kamu habis sebelum akhir tahun !')
            break
        minggu += 1

    else:
        print(f'Saldo akhir bulan ke-{bulan} : Rp.{saldo:,.0f}\n')
        continue
    break

else:
    print(f'\n Nobita berhasil menabung Rp{saldo} dari total Rp {saldo_awal}')

```
