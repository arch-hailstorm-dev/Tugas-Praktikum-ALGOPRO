```python
#   Variabel / Data Tingkat Risiko Asuransi dan Premi

usia_pengemudi  = int(input('Masukkan usia pengemudi'))
jenis_kendaraan = input('Masukkan jenis kendaraan (Mobil/Motor')
nilai_kendaraan = float(input('Masukkan nilai kendaraan Rp:'))
area_penggunaan = input('Masukkan area penggunaan (Perkotaan/Pedesaan):')
sistem_pengaman = input('Masukkan sistem pengaman (Ya/Tidak):')


#   Menggabungkan variabel dengan kondisinya


#   Usia Pengemudi

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = 'tinggi'

else:
    risiko = 'rendah'


#   Jenis Kendaraan

if jenis_kendaraan == 'motor':
    risiko_akhir = 'Sedang'

else:
    risiko_akhir = 'Rendah'


#   Nilai Kendaraan (Menghitung Premi)

if nilai_kendaraan == 'Tinggi':
    premi = 0.1 * nilai_kendaraan

elif nilai_kendaraan == 'Sedang':
    premi = 0.07 * nilai_kendaraan

else:
    premi = 0.05 * nilai_kendaraan


#   Area Penggunaan

if area_penggunaan == 'perkotaan':
    if risiko_akhir == 'rendah':
        risiko_akhir = 'Sedang'

    elif risiko_akhir == 'Sedang':
        risiko_akhir = 'Tinggi'

#   Sistem Pengaman

if sistem_pengaman == 'tidak':
    if risiko_akhir == 'rendah':
        risiko_akhir = 'sedang'

    elif risiko_akhir == 'Sedang':
        risiko_akhir = 'Tinggi'

#   Pesan Doraemon

pesan = 'Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!' \
    if risiko_akhir == 'Tinggi' else'Risiko Kendaraan masih aman menurut Doraemon'


#   Hasil Akhir DORABIKE

judul_2 = 'Smart Vehicle Insurance Analyzer by Arch x Hailstorm (Abyy)'
print('='*60)
print(f'{judul_2:^60}')
print('='*60)
print(sep='......')

print(f'1. Usia Pengemudi      :', usia_pengemudi)
print(f'2. Jenis Kendaraan     :', jenis_kendaraan)
print(f'3. Nilai Kendaraan     : Rp{nilai_kendaraan:,.0f}')
print(f'4. Area Penggunaan     :', area_penggunaan)
print(f'5. Sistem Pengaman     : ', sistem_pengaman)
print(sep='.....')
nilai_akhir = 'Hasil Analisis DORABIKE'
print('-'*40)
print(f'{nilai_akhir:^40}')
print('-'*40)
print(f'1. Kategori Risiko : {risiko_akhir.capitalize()}')
print(f'2. Premi Asuransi  : Rp{premi:,.0f}')
print(f'3. Pesan Doraemon  : {pesan}')


```
