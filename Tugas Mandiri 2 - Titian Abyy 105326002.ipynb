```python
#   Hitung BMI Giant

print('=' * 53)
berat_badan = 97.5  # kg
tinggi_badan = 1.75 # meter

bmi = berat_badan / (tinggi_badan ** 2)   # BMI adalah Body Mass Index atau bisa disebut juga dengan IMT (Indeks Massa Tubuh)
bmi_pembulatan = round(bmi, 2)   # Pembulatan BMI Giant

print(f'Nilai Body Mass Index Giant asli = {bmi}')
print(f'Nilai Body Mass Index Giant yang dibulatkan = {bmi_pembulatan}')
print('=' * 53)
```

```python
#   Target minum Giant

print('='*35)
target_minum_per_hari = 2000 # mililiter (ml)
minum_per_botol = 350        # mililiter per botol (ml)

#   True Division

hasil_true_division = target_minum_per_hari / minum_per_botol

print('Hasil asli =', hasil_true_division, 'ml')
print('='*35)

#   Floor Division

hasil_floor_division = target_minum_per_hari // minum_per_botol

print('Hasil Botol Penuh =', hasil_floor_division, 'ml')
print('='*35)

#   Modulo

hasil_modulo = target_minum_per_hari % minum_per_botol

print('Hasil Sisa air botol =', hasil_modulo, 'ml')
print('='*35)
```

```python
#   Kebutuhan Kalori Harian Giant

print('='*72)
import math
kalori_harian = 2100.5  #kalori per hari Giant
kalori_floor = math.floor(kalori_harian)
kalori_ceil = math.ceil(kalori_harian)

print('Kebutuhan Kalori Harian Giant =', kalori_harian, 'kalori')
print('Kebutuhan Kalori Harian Giant setelah dibulatkan ke bawah =', kalori_floor, 'kalori')
print('Kebutuhan Kalori Harian Giant setelah dibulatkan ke atas =', kalori_ceil, 'kalori')
print('='*72)

#  -------selesai-------
```
