```python
#   1. Input / Mengisi Nama Makanan, Nama Waifu, Kata Sifat

nama_makanan = input("masukkan makanan favorit: ")
nama_waifu = input("masukkan nama waifu: ") #contoh : mahiru, lyudmila, zerotwo
kata_sifat = input("masukkan sifat lu: ") #contoh : baik hati, bijaksana, dan keren

print(nama_makanan)
print(nama_waifu)
print(kata_sifat)
```

```python
#   3. Modal awal 3 orang

Doraemon = 555000
Nobita = 333000
Shizuka = 222000
```

```python
#   4. Menghitung total modal

total_modal = Doraemon + Nobita + Shizuka

print ('Total modal awal:' , total_modal)
```

```python
#   5. Penjualan dan Pendapatan dari Toko Dorayaki

#   Dorayaki
jumlah_dorayaki = 350
harga_dorayaki = 5000

#   Menghitung Pendapatan
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print('Total Pendapatan:' , total_pendapatan)
```

```python
#   6. Menghitung keuntungan dari Penjualan Dorayaki

#   Keuntungan
keuntungan = total_pendapatan - total_modal

print('Keuntungan:' , keuntungan)
```
